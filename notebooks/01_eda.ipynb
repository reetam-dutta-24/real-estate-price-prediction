```python
import numpy as np

def haversine_distance(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c

seattle_lat, seattle_lon = 47.6062, -122.3321
df['distance_to_seattle'] = haversine_distance(df['lat'], df['long'], seattle_lat, seattle_lon)
```

```python
# sale_quarter
df['sale_quarter'] = df['date'].dt.quarter

# sale_day_of_week (0=Monday, 6=Sunday)
df['sale_day_of_week'] = df['date'].dt.dayofweek

# is_peak_season (May-Aug, historically hottest US real estate months)
df['is_peak_season'] = df['sale_month'].isin([5,6,7,8]).astype(int)

# decade_built
df['decade_built'] = (df['yr_built'] // 10) * 10

# renovation_decade (0 if never renovated)
df['renovation_decade'] = np.where(
    df['yr_renovated'] > 0,
    (df['yr_renovated'] // 10) * 10,
    0
)

# cyclical encoding for sale_month
df['sale_month_sin'] = np.sin(2 * np.pi * df['sale_month'] / 12)
df['sale_month_cos'] = np.cos(2 * np.pi * df['sale_month'] / 12)
```

```python
df[['sale_quarter','sale_day_of_week','is_peak_season','decade_built','renovation_decade','sale_month_sin','sale_month_cos']].head()
```

```python
# above_ratio — proportion of house above ground vs total living space
df['above_ratio'] = df['sqft_above'] / df['sqft_living']

# how this house compares to its immediate neighbors
df['living_sqft_diff_from_neighbors'] = df['sqft_living'] - df['sqft_living15']
df['lot_sqft_diff_from_neighbors'] = df['sqft_lot'] - df['sqft_lot15']

# rough total room count
df['total_rooms_estimate'] = df['bedrooms'] + df['bathrooms']
```

```python
df[['above_ratio','living_sqft_diff_from_neighbors','lot_sqft_diff_from_neighbors','total_rooms_estimate']].describe()
```

```python
#Luxary
# view_binary — has any view at all, regardless of intensity (0-4 scale)
df['view_binary'] = (df['view'] > 0).astype(int)

# is_luxury — composite flag: waterfront OR top-tier grade OR strong view
df['is_luxury'] = ((df['waterfront'] == 1) | (df['grade'] >= 11) | (df['view'] >= 3)).astype(int)

# luxury_score — weighted composite, normalized inputs combined into one continuous signal
df['luxury_score'] = (
    df['grade'] * 1.0 +
    df['view'] * 2.0 +
    df['waterfront'] * 5.0 +
    df['condition'] * 0.5
)
```

```python
distance_cols = [c for c in df.columns if c.startswith('distance_to_')]
print(f"Total distance features: {len(distance_cols)}")
df[distance_cols].describe().T.head(10)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

scaler = StandardScaler()
distance_scaled = scaler.fit_transform(df[distance_cols])
```

```python
pca_full = PCA()
pca_full.fit(distance_scaled)

import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(range(1, len(pca_full.explained_variance_ratio_)+1), 
         pca_full.explained_variance_ratio_.cumsum(), marker='o')
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA — How Many Components Do We Need?")
plt.axhline(y=0.95, color='r', linestyle='--', label='95% variance threshold')
plt.legend()
plt.savefig("../reports/figures/pca_explained_variance.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
pca_final = PCA(n_components=3)
distance_pca = pca_final.fit_transform(distance_scaled)

df['amenity_proximity_pc1'] = distance_pca[:, 0]
df['amenity_proximity_pc2'] = distance_pca[:, 1]
df['amenity_proximity_pc3'] = distance_pca[:, 2]

print("Variance explained by each component:", pca_final.explained_variance_ratio_)
print("Total variance retained:", pca_final.explained_variance_ratio_.sum())
```
