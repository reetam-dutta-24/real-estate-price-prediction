```python
df = df.drop(columns=['Unnamed: 0', 'id'])
```

```python
df.shape
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df["date"].head(10)
```

```python
df['sale_year'] = df['date'].dt.year
df['sale_month'] = df['date'].dt.month
```

```python
df["sale_year"].head(10)
```

```python
df["sale_month"].head(5)
```

```python
#We need to use feature engineer to make sure we convert the date data to year of sale and month as models cant compute from raw date data, we use sale year to determine the age of the property and month is also important for seasonal variation
```

```python
df[df['bedrooms'].isnull() | df['bathrooms'].isnull()]
```

```python
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].median())
df['bathrooms'] = df['bathrooms'].fillna(df['bathrooms'].median())
```

```python
df.duplicated().sum()
```

```python
df[df['bedrooms'] == 0]
df[df['bathrooms'] == 0]
df[df['price'] <= 0]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18,5))
sns.boxplot(y=df['price'], ax=axes[0])
sns.boxplot(y=df['sqft_living'], ax=axes[1])
sns.boxplot(y=df['sqft_lot'], ax=axes[2])
plt.tight_layout()
plt.savefig("../reports/figures/outlier_boxplots.png", dpi=150, bbox_inches='tight')
plt.show()
```
